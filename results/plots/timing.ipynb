```python
import seaborn as sns
import pandas as pd

data = pd.read_csv("../experiments/ttft/ttft_fixed_seq.csv")

ax = sns.lineplot(data, x='num_loras', y='time')

ax.set(xlabel='Number of LoRAS loaded', ylabel='ttft [s]')
ax.set_title("Time-to-first-token for ACLoRA for 64 input-tokens")
```

```python
import numpy as np

data = pd.read_csv("../experiments/ttft/ttft_fixed_seq_no_loading.csv")
data_vanilla = pd.read_csv("../experiments/ttft/ttft_vanilla.csv")


ax = sns.lineplot(data, x='num_loras', y='time', marker='o')

ax.set(xlabel='Number of active LoRAS', ylabel='ttft [s]')
ax.set_title("Time-to-first-token for ACLoRA for 89 input-tokens")

set(list(data_vanilla['context_size']))
data_vanilla_mean_10535 = data_vanilla[data_vanilla['context_size']==10535]
mean_value_data_10535 = np.array(data_vanilla_mean_10535['time']).mean()
ax.axhline(y=mean_value_data_10535, color='gray', linestyle='--', label='Vanilla with 10k context')
ax.text(x=1, y=mean_value_data_10535+0.03, s='Vanilla with 10k context', color='gray', ha='left', va='bottom')

data_vanilla_mean_5285 = data_vanilla[data_vanilla['context_size']==5285]
mean_value_data_5285 = np.array(data_vanilla_mean_5285['time']).mean()
ax.axhline(y=mean_value_data_5285, color='gray', linestyle='--', label='Vanilla with 5k context')
ax.text(x=1, y=mean_value_data_5285+0.03, s='Vanilla with 5k context', color='gray', ha='left', va='bottom')


data_vanilla_mean_5285 = data_vanilla[data_vanilla['context_size']==259]
mean_value_data_5285 = np.array(data_vanilla_mean_5285['time']).mean()
ax.axhline(y=mean_value_data_5285, color='gray', linestyle='--', label='Vanilla with 260 context')
ax.text(x=10, y=mean_value_data_5285+0.03, s='Vanilla with 260 context', color='gray', ha='left', va='bottom')
```
